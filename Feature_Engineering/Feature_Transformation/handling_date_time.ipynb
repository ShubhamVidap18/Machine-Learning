```python
# day of week

date['date_dow'] = date['date'].dt.dayofweek
```

```python
# day pf week name

date['dow_name'] = date['date'].dt.day_name()
```

```python
# is weekend?
date['is_weekend'] = np.where(date['dow_name'].isin(['Sunday', 'Saturday']), 1, 0)
date.drop(columns = ['product_id', 'city_id', 'orders']).head()
```
